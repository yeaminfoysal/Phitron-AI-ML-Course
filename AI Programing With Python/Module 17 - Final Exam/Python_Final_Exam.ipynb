# 0) Starter Code (Run First)

```python
# You must run this code and use this dataset for all questions
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
num_students = 50

# Student data
student_ids = np.arange(201, 201 + num_students)
student_ages = np.random.randint(15, 22, num_students)
student_scores = np.random.randint(40, 100, num_students)
student_attendance = np.random.randint(60, 100, num_students)  # percentage

departments_list = [
    "Science",
    "Commerce",
    "Arts",
    "Computer Science",
    "Mathematics",
    "English",
    "Physics",
    "Chemistry",
    "Biology",
    "Economics"
]

student_departments = np.random.choice(departments_list, num_students)

grade_conditions = [
    student_scores >= 80,
    (student_scores >= 65) & (student_scores < 80),
    (student_scores >= 50) & (student_scores < 65),
    student_scores < 50
]
grade_values = ["A", "B", "C", "F"]
student_grades = np.select(grade_conditions, grade_values, default="F")

# Create DataFrame
data = pd.DataFrame({
    "student_id": student_ids,
    "student_age": student_ages,
    "score": student_scores,
    "attendance": student_attendance,
    "department": student_departments,
    "grade": student_grades
})

# Save dataset
data.to_csv("school_data.csv", index=False)
```

# Question 01: [ Marks 10 ]

Use NumPy arrays from the score and attendance columns. Print both arrays and display:

* Data type
* Number of dimensions
* Shape
* Total number of elements

# Write Answer 01:


```python
from pandas.io.parsers import read_csv
import numpy as np
import pandas as pd

data= read_csv('school_data.csv')

score = np.array(data['score'])
attendance = np.array(data['attendance'])

print("Score Array:")
print(score)

print("\nAttendance Array:")
print(attendance)

print("\nScore Array Information:")
print("Data type",score.dtype)
print("Number of dimensions",score.ndim)
print("Shape",score.shape)
print("Total number of elements",score.size)

print("\nAttendance Array Information:")
print("Data type",attendance.dtype)
print("Number of dimensions",attendance.ndim)
print("Shape",attendance.shape)
print("Total number of elements",attendance.size)

```

```python
max_score = np.max(score)
min_score = np.min(score)
avg_score = np.mean(score)
avg_attendance = np.mean(attendance)

print("Maximum Score:", max_score)
print("Minimum Score:", min_score)
print("Average Score:", avg_score)
print("Average Attendance:", avg_attendance)

filtered_students = np.where(score > avg_score)

filtered_score = score[filtered_students]
print("\nFiltered Score Array:")
print(filtered_score)
```


# Question 04: [ Marks 10 ]

##
Using .loc:

* Increase the score by 10% for students in "Computer Science" and "Mathematics" departments
* Increase the score by 5% for all other departments
* Cap the score at 100 (no score should exceed 100 after the raise)

Then print a DataFrame that includes:

* student_id, department, previous_score, new_score, score_increase


### Write Answer 04:

```python
data2= data.copy()
data2['previous_score'] = data2['score']
data2['score']= data2['score'].astype(float)

data2.loc[data2['department'].isin(['Mathematics', 'Computer Science']), 'score'] *= 1.1
data2.loc[~data2['department'].isin(['Computer Science', 'Mathematics']), 'score'] *= 1.05
data2.loc[data2['score'] > 100, 'score'] = 100

data2['score_increase'] = data2['score'] - data2['previous_score']

final_data = data2[['student_id', 'department', 'previous_score', 'score', 'score_increase']].rename(columns={'score':'new_score'})

display(final_data)
```

# Question 05: [ Marks 10 ]

Using Matplotlib, create a bar chart that shows the average score per department.

* X-axis: Department names
* Y-axis: Average score
* Add a title: "Average Score by Department"
* Display the chart

## Answer 05:

```python
import matplotlib.pyplot as plt
avg_scores = data.groupby('department')['score'].mean()
plt.bar(avg_scores.index, avg_scores.values)
plt.xlabel('Department')
plt.ylabel('Average Score')
plt.title('Average Score by Department')
plt.xticks(rotation=45, ha='right')
plt.show()
```
