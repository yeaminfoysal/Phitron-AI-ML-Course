# List as Stack

```python
stack =[]

# stack e push
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)


print(f"top element : {stack[-1]}")
# removing the top element
stack.pop()

# top element
print(f"top element : {stack[-1]}")
```

# List as Queue

```python
queue = []

# insertion
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)


#remove
print(queue.pop(0))


# accessing the front element

print(f"front e {queue[0]}")

queue.pop(0)

print(f"front e {queue[0]}")
```

# list comprehension

```python
even = []

# Normal way
for i in range(101):
    if i%2==0:
        even.append(i)

print("even_number",even)

# list comprehension
odd_number = [x for x in range(10) if x % 2 != 0]
print("odd_number", odd_number)

random_list = [x+10 for x in range(1,101) if x%2!=0]

print(random_list)
```

```python
fruits = ['APPLE' , 'ORange' , 'liCHi']

# Normal way
lower_fruits=[]
for fruit in fruits:
  lower_fruits.append(fruit.lower())

print(lower_fruits)

# list comprehension
upper_fruits=[fruit.upper() for fruit in fruits]
print(upper_fruits)
```
