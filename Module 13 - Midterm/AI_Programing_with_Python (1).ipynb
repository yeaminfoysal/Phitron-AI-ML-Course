## 0) Starter Code (Run First)

```python
import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

## Q1.

```python
print("Employee Ages: ", emp_ages)
print("Employee Income: ", emp_income)
print()

print("----- emp_ages -----")
print("Data type          :", emp_ages.dtype)
print("Dimensions (ndim)  :", emp_ages.ndim)
print("Shape              :", emp_ages.shape)
print("Total elements     :", emp_ages.size)
print()

print("----- emp_income -----")
print("Data type          :", emp_income.dtype)
print("Dimensions (ndim)  :", emp_income.ndim)
print("Shape              :", emp_income.shape)
print("Total elements     :", emp_income.size)
```

## Q2.

```python
max_salary = np.max(emp_income)
min_salary = np.min(emp_income)
avg_salary = np.mean(emp_income)
avg_age = np.mean(emp_ages)

print("Maximum Salary: ", max_salary)
print("Minimum Salary: ", min_salary)
print("Average Salary: ", round(avg_salary, 2))
print("Average Age: ", round(avg_age, 2))
```

## Q3.

```python
combined_condition = (emp_ages >= 25) & (emp_ages <= 40) & (emp_income > 50000)

filtered_ages = emp_ages[combined_condition]
filtered_salary = emp_income[combined_condition]

print("Filtered ages  :", filtered_ages)
print("Their salaries :", filtered_salary)
print("Number of employees satisfy this condition:", np.count_nonzero(combined_condition))
```

## Q4

```python
average_age_used = np.mean(emp_ages)

updated_ages = np.where(emp_ages < average_age_used, emp_ages + 4, emp_ages + 2)

print("Original array: ", emp_ages)
print("Updated array: ", updated_ages)
print("Average age used for condition", round(average_age_used, 2))
```

## Q5

```python
tech_roles = ["Software Developer", "Backend Developer"]

is_tech = np.isin(emp_roles, tech_roles)

new_salary = np.where(is_tech, emp_income * 1.08, emp_income)

old_total = np.sum(emp_income)
new_total = np.sum(new_salary)
difference = new_total - old_total

print("Number of tech employees :", np.count_nonzero(is_tech))
print("Old total salary expense :", round(old_total, 2))
print("New total salary expense :", round(new_total, 2))
print("Overall increase         :", round(difference, 2))
```

## Q10.

```python
result = df[["role"]].copy()
result["previous_salary"] = df["emp_salary"]

dev_mask = df["role"].isin(["Software Developer", "Backend Developer"])

result.loc[dev_mask, "new_salary"]  = (df.loc[dev_mask, "emp_salary"] * 1.12).round(2)
result.loc[~dev_mask, "new_salary"] = (df.loc[~dev_mask, "emp_salary"] * 1.05).round(2)

result["increase_of_salary"] = (result["new_salary"] - result["previous_salary"]).round(2)

display(result)
```
